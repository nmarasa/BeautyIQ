```python
import pandas as pd
import sqlite3
```

```python
products = pd.read_csv("../data/processed/products_clean.csv")
reviews = pd.read_csv("../data/processed/skincare_reviews_clean.csv")
```

```python
conn = sqlite3.connect("../data/processed/beautyiq.db")
```

```python
products.to_sql(
    "products",
    conn,
    if_exists="replace",
    index=False
)
```

```python
reviews.to_sql(
    "reviews",
    conn,
    if_exists="replace",
    index=False,
    chunksize=50000
)
```

```python
pd.read_sql_query(
    "SELECT COUNT(*) AS product_count FROM products",
    conn
)
```

```python
pd.read_sql_query(
    "SELECT COUNT(*) AS review_count FROM reviews",
    conn
)
```

```python
conn.close()
```
