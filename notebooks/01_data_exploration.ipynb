# initial Review Data Findings

- `reviews_0-250.csv` contains 602,130 reviews.
- represents 250 unique product IDs across 60 brands.
- review dates range from 2008 to 2023.
- ratings range from 1 to 5, with 5-star reviews being the most common.
- `product_id` is available, allowing review data to be connected to `product_info.csv`.
- review text is almost completely populated.

```python
review_files = [
    "../data/raw/reviews_0-250.csv",
    "../data/raw/reviews_250-500.csv",
    "../data/raw/reviews_500-750.csv",
    "../data/raw/reviews_750-1250.csv",
    "../data/raw/reviews_1250-end.csv"
]
```

```python
for file in review_files:
    df = pd.read_csv(file, usecols=["product_id", "brand_name"])
    
    print(file)
    print("Rows:", len(df))
    print("Unique products:", df["product_id"].nunique())
    print("Unique brands:", df["brand_name"].nunique())
    print()
```

```python
review_dfs = []

for file in review_files:
    temp = pd.read_csv(file)
    review_dfs.append(temp)

reviews = pd.concat(review_dfs, ignore_index=True)
```

## core business questions

1. Which beauty categories and brands differ most in price, rating, and customer interest across the full Sephora catalog?
2. Within skincare, which brands and products show the strongest customer demand?
3. Which skincare products have high customer interest but relatively weak ratings?
4. What complaints appear most frequently in negative skincare reviews?
5. What qualities appear most frequently in positive skincare reviews?
6. Which combinations of demand, price, rating, and review themes suggest potential skincare product opportunities?
